```python
import pandas as pd
```

```python
df=pd.read_csv("jobs_features.csv")
```

```python
df["Role"]=df["Role"].str.strip()
df["Role"]=df["Role"].str.replace(r"\s+"," ",regex=True)
```

```python
df["Role"].value_counts()
```

```python
import re
```

```python
def clean_text(text):
    text=str(text).lower()
    text=re.sub(r"http\S+"," ",text)
    text=re.sub(r"[^a-zA-Z0-9]"," ",text)
    text=re.sub(r"\s+"," ",text)
    return text
```

```python
df["Features"]=df["Features"].apply(clean_text)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoder=LabelEncoder()
```

```python
df["Role"]=encoder.fit_transform(df["Role"])
```

```python
x=df["Features"]
y=df["Role"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf=TfidfVectorizer(stop_words="english")
```

```python
X_train_tfidf=tfidf.fit_transform(x_train)
X_test_tfidf=tfidf.transform(x_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model=RandomForestClassifier(n_estimators=100,random_state=42,n_jobs=-1)
```

```python
model.fit(X_train_tfidf,y_train)
```

```python
y_pred=model.predict(X_test_tfidf)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy=accuracy_score(y_test,y_pred)
print("Accuracy:",accuracy)
```

```python
resume = ["Python SQL Machine Learning Deep Learning Pandas NumPy Data Analysis"]
resume_tfidf = tfidf.transform(resume)

prediction = model.predict(resume_tfidf)

predicted_role = encoder.inverse_transform(prediction)

print("Predicted Role:", predicted_role[0])
```
